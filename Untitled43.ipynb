```python
import pandas as pd
docs = [
    "Machine learning models like random forest and logistic regression perform well on tabular data.",
    "Deep learning, CNNs, and computer vision pipelines process video frames efficiently.",
    "Natural language processing uses transformer models for text classification and sentiment analysis.",
    "AI agents and workflow automation tools streamline daily tasks.",
]
```

```python
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import CountVectorizer
vectorizer_model = CountVectorizer(stop_words="english", ngram_range=(1, 2))
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
topic_model = BERTopic(
    embedding_model=embedding_model,
    vectorizer_model=vectorizer_model,
    language="english",
    calculate_probabilities=True,
)
```

```python
from bertopic import BERTopic
from hdbscan import HDBSCAN
from sentence_transformers import SentenceTransformer
from umap import UMAP

# 1. UMAP model configure karein (chote dataset ke liye)
umap_model = UMAP(
    n_neighbors=2, n_components=2, min_dist=0.0, metric="cosine"
)

# 2. HDBSCAN model configure karein (min_cluster_size aur min_samples ko kam kar dein)
hdbscan_model = HDBSCAN(
    min_cluster_size=2,
    min_samples=1,
    metric="euclidean",
    prediction_data=True,
)

# 3. Embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# 4. Sab models ko BERTopic mein pass kar dein
topic_model = BERTopic(
    embedding_model=embedding_model,
    umap_model=umap_model,
    hdbscan_model=hdbscan_model,
)

# 5. Ab fit karein (Ab error nahi aayega!)
topics, probabilities = topic_model.fit_transform(docs)

print(topic_model.get_topic_info())
```

```python
from transformers import pipeline

# Sentiment analysis pipeline load karein
sentiment_pipeline = pipeline(
    "sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Saare documents par run karein (batch processing ke liye list comprehension)
sentiments = []
for doc in docs:
  # Truncate text if too long for the model token limit
  res = sentiment_pipeline(doc[:512])[0]
  sentiments.append(res["label"])  # 'POSITIVE' ya 'NEGATIVE'
```

```python
from sklearn.metrics import accuracy_score, classification_report
true_labels = [
    "POSITIVE",
    "POSITIVE",
    "POSITIVE",
    "NEUTRAL",
]
pred_labels = sentiments[: len(true_labels)]
accuracy = accuracy_score(true_labels, pred_labels)
print(f"Sentiment Analysis Accuracy: {accuracy * 100:.2f}%")
print(classification_report(true_labels, pred_labels, zero_division=0))
```

```python
import chromadb
client = chromadb.Client()
collection = client.get_or_create_collection(name="nlp_metadata_corpus")
embeddings = embedding_model.encode(docs)
metadatas = [
    {
        "topic": int(topics[i]),
        "sentiment": str(sentiments[i]),
        "document_id": i,
    }
    for i in range(len(docs))
]
collection.add(
    documents=docs,
    embeddings=embeddings.tolist(),
    metadatas=metadatas,
    ids=[f"doc_{i}" for i in range(len(docs))],
)
print("Metadata successfully integrated into ChromaDB!")
```

```python
query_text = "machine learning models"
query_embedding = embedding_model.encode([query_text]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=3,
    where={
        "$and": [{"topic": 0}, {"sentiment": "POSITIVE"}]
    },
)

print("Filtered Results:", results["documents"])
```

```python
import pandas as pd
import chromadb
from bertopic import BERTopic
from transformers import pipeline
from sentence_transformers import SentenceTransformer
from sklearn.metrics import accuracy_score, classification_report
client = chromadb.PersistentClient(path="./chroma_db")
collection = client.get_collection(
    name="research_documents"
)
data = collection.get(
    include=["documents", "metadatas"]
)
ids = data["ids"]
documents = data["documents"]
metadatas = data["metadatas"]
valid_indices = [
    i for i, doc in enumerate(documents)
    if doc and len(doc.split()) >= 30
]
valid_documents = [
    documents[i] for i in valid_indices
]
if len(valid_documents) < 10:
    raise ValueError(
        "Not enough valid documents for topic modeling."
    )
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
topic_model = BERTopic(
    embedding_model=embedding_model,
    min_topic_size=10,
    nr_topics="auto",
    verbose=True
)
topics, probabilities = topic_model.fit_transform(
    valid_documents
)
topic_info = topic_model.get_topic_info()
print("\nDISCOVERED TOPICS")
print(topic_info)
topic_names = {}
for _, row in topic_info.iterrows():
    topic_names[int(row["Topic"])] = str(
        row["Name"]
    )
for topic_id in sorted(set(topics)):
    if topic_id == -1:
        continue
    print(f"\nTOPIC {topic_id}")
    print("Keywords:", topic_model.get_topic(topic_id))
    indices = [
        i for i, t in enumerate(topics)
        if t == topic_id
    ][:3]
    for index in indices:
        print("Document:", valid_documents[index][:500])
sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
def analyze_sentiment(text):
    if not text or len(text.split()) < 5:
        return "insufficient_text", 0.0

    result = sentiment_analyzer(
        text[:1500],
        truncation=True
    )[0]
    return (
        result["label"].lower(),
        float(result["score"])
    )
nlp_metadata = {}
for i in range(len(documents)):
    nlp_metadata[i] = {
        "topic_id": -1,
        "topic_name": "short_or_unassigned",
    }
for position, original_index in enumerate(valid_indices):
    topic_id = int(topics[position])

    nlp_metadata[original_index] = {
        "topic_id": topic_id,
        "topic_name": topic_names.get(
            topic_id, "unassigned"
        )
    }
for start in range(0, len(documents), 32):
    batch_end = min(start + 32, len(documents))
    batch_texts = documents[start:batch_end]
    batch_results = sentiment_analyzer(
        [
            text[:1500] if text and len(text.split()) >= 5
            else "This document has insufficient text."
            for text in batch_texts
        ],
        truncation=True
    )
    for offset, result in enumerate(batch_results):
        index = start + offset
        if not documents[index] or len(documents[index].split()) < 5:
            label = "insufficient_text"
            score = 0.0
        else:
            label = result["label"].lower()
            score = float(result["score"])
        nlp_metadata[index]["sentiment"] = label
        nlp_metadata[index]["sentiment_score"] = score
for start in range(0, len(ids), 500):
    end = min(start + 500, len(ids))
    batch_ids = ids[start:end]
    updated_metadata = []
    for index in range(start, end):
        existing = dict(metadatas[index] or {})
        existing.update(nlp_metadata[index])
        updated_metadata.append(existing)
    collection.update(
        ids=batch_ids,
        metadatas=updated_metadata
    )
print("\nNLP METADATA SUCCESSFULLY SAVED!")
print("Total documents:", len(documents))
test_data = pd.read_csv(
    "data/processed/manual_labels.csv"
)
predictions = [
    analyze_sentiment(text)[0]
    for text in test_data["text"]
]
gold_labels = [
    str(label).lower()
    for label in test_data["gold_label"]
]
accuracy = accuracy_score(
    gold_labels,
    predictions
)
print("\nSENTIMENT EVALUATION")
print("Accuracy:", round(accuracy * 100, 2), "%")

print(
    classification_report(
        gold_labels,
        predictions,
        zero_division=0
    )
)
```
